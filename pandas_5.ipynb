```python
import pandas as pd
import numpy as np
import random
```

```python
df_exams = pd.read_csv("StudentsPerformance.csv")
```

```python
#first raws
df_exams.head()
```

#### Create Index

```python
# creating non repetitive values for the index
new_index = np.arange(0, 1000)
new_index
```

```python
# shuffling the indexes
random.shuffle(new_index)
new_index
```

```python
# create a new column with the new index
df_exams["new_index"] = new_index
df_exams
```

### Set Index

```python
# set new_index as index
df_exams.set_index("new_index", inplace=True)
df_exams
```

#### Sort Index


```python
# sort dataframe by new index
df_exams.sort_index()
```

```python
# sort in descending order by index
df_exams.sort_index(ascending=False)
```

#### Renaming Columns

```python
# rename column and overwrite th values of the dataframe
df_exams = df_exams.rename(columns={"gender":"Gender"})
```

```python
# rename multiple columns
df_exams.rename(columns={"math score":"MS", "reading score":"RS", "writing score":"WS"}, inplace=True)
```

```python
# show dataframe
df_exams.head()
```

### Renaming Index

```python
# rename index 0,1,and 2 and update the dataframe
df_exams.rename(index={0:"A", 1:"B", 2:"C"}, inplace=True)
```
