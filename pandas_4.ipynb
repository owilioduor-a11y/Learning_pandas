```python
import pandas as pd 
```

```python
# reading csv file
df_exams = pd.read_csv("StudentsPerformance.csv")
df_exams.head()
```

### Operations in columns

```python
# select a column and calculate total sum
df_exams["math score"].sum()
```

```python
# count, mean, std, max and min
df_exams["math score"].count()
df_exams["math score"].mean()
df_exams["math score"].std()
df_exams["math score"].max()
df_exams["math score"].min()
```

```python
# calculate using describe()
df_exams.describe()
```

### Operations in rows

```python
# calculating the sum in a row
df_exams["math score"] + df_exams["reading score"] + df_exams["writing score"]
```

```python
# calculating the average score and assigning the result to new column
df_exams["average"] = (df_exams["math score"] + df_exams["reading score"] + df_exams["writing score"])/3
```

### Value count

```python
# Counting the gender element
# len function
len(df_exams['gender'])
# .count() method 
df_exams['gender'].count()
```

```python
# counting gender elements by category
df_exams['gender'].value_counts()
```

```python
# return the reltive frequency (divide all values by sum of the values)
df_exams['gender'].value_counts(normalize=True)
```

```python
# counting "parental level of education" elements by category
df_exams['parental level of education'].value_counts()
```

```python
#  return to the frequency and round to 2 decimals
df_exams['parental level of education'].value_counts(normalize=
True).round(2)
```

#### Sort a dataframe

```python
# sort by one column
df_exams.sort_values('math score')
```

```python
# sort in descending by one column
df_exams.sort_values("math score", ascending=False)
```

```python
# sort in descending order by multiple columns
df_exams.sort_values(["math score", "reading score"], ascending=False)
```

```python
# sort by multiple columns and update the dataframe
df_exams.sort_values(["math score", "reading score"], ascending=False, inplace=True)
```

```python
# sort in descending with a key function
df_exams.sort_values(["race/ethnicity"], ascending=True, key=lambda a:a.str.lower())
```
