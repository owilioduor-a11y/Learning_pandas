### Adding new columns using assign()
#### When to use assign()
###### 1. when adding multiple columns in a single line of code
###### 2. when you need to overwrite the value of existing column

```python
import numpy as np
```

```python
# creating random number for our new score columns
score1 = np.random.randint(1,100, size=1000)
score2 = np.random.randint(1,100, size=1000)
```

```python
# creating series using random numbers
series1 = pd.Series(score1, index=np.arange(0, 1000))
series2 = pd.Series(score2, index=np.arange(0 , 1000))
```

```python
# using assign() to add multiple columns
df_exams = df_exams.assign(score1=series1, score2=series2)
```

```python
df_exams
```

### adding column using insert()

```python
# using insert() to add column at a specific position
df_exams.insert(1, "test", series1)
df_exams
```
